```python
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.messages import HumanMessage,AIMessage
```

```python
load_dotenv()
model = init_chat_model(
    model = "kimi-k3",
    model_provider="openai",
    api_key = os.getenv("OPEN_API_KEY"),
    base_url = os.getenv("OPEN_BASE_URL"),
)
agent = create_agent(model)
# messages = HumanMessage([
#      # {"type": "image","url": "https://www.sensenova.cn/marketing-home/showcase-hero.png"},
#     {"type":"text","text":"识别一下图片内容"}
# ])
result = agent.invoke(
    {"messages":[HumanMessage(content="你好啊")]}
)

messageList = result["messages"]
for ms in messageList:
    if isinstance(ms, AIMessage):
        print(f"AI：{ms.content}")
    else:
        print(f"用户：{ms.content}")
```
